````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# 🤖 Single Agent Pipeline - Solution

def agent(query: str):
    try:
        query_lower = query.lower().strip()

        # Conditional Routing
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "", 1).strip()

            return {
                "type": "calculation",
                "result": calculator(expression)
            }

        elif "keywords" in query_lower:
            text = query_lower.replace("extract keywords from", "", 1).strip()

            return {
                "type": "keywords",
                "result": extract_keywords(text)
            }

        else:
            return {
                "type": "general",
                "result": f"Direct response: {query}"
            }

    except Exception as e:
        return {
            "type": "error",
            "result": str(e)
        }
````

````python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)


def word_count(text: str) -> int:
    """Count the number of words in the given text."""
    try:
        return len(text.split())
    except Exception:
        return 0


def agent(query: str):

    try:
        if not query or not query.strip():
            logging.warning("Empty query received")

            return {
                "type": "error",
                "result": "Query cannot be empty"
            }

        query_lower = query.lower().strip()

        logging.info(f"Received query: {query}")

        math_keywords = [
            "calculate",
            "solve",
            "addition",
            "subtraction",
            "multiply",
            "multiplication",
            "divide",
            "division"
        ]

        math_symbols = ["+", "-", "*", "/"]

        is_math = (
            any(word in query_lower for word in math_keywords)
            or (
                any(symbol in query for symbol in math_symbols)
                and any(char.isdigit() for char in query)
            )
        )

        if is_math:

            logging.info("Routing query to Calculator Tool")

            expression = query_lower

            for word in [
                "calculate",
                "solve",
                "what is",
                "please calculate"
            ]:
                expression = expression.replace(word, "")

            expression = expression.strip()

            result = calculator(expression)

            if result == "Error in calculation":
                logging.error("Calculator error")

                return {
                    "type": "error",
                    "result": result
                }

            return {
                "type": "calculation",
                "result": result
            }


        keyword_phrases = [
            "keywords",
            "keyword",
            "extract keywords",
            "find keywords"
        ]

        if any(phrase in query_lower for phrase in keyword_phrases):

            logging.info("Routing query to Keyword Extraction Tool")

            text = query

            for phrase in [
                "extract keywords from",
                "extract keyword from",
                "find keywords from",
                "find keyword from"
            ]:
                text = text.lower().replace(phrase, "")

            result = extract_keywords(text.strip())

            return {
                "type": "keywords",
                "result": result
            }



        word_count_phrases = [
            "word count",
            "count words",
            "number of words"
        ]

        if any(phrase in query_lower for phrase in word_count_phrases):

            logging.info("Routing query to Word Count Tool")

            text = query_lower

            for phrase in word_count_phrases:
                text = text.replace(phrase, "")

            result = word_count(text.strip())

            return {
                "type": "word_count",
                "result": result
            }

        logging.info("Routing query to General Response")

        return {
            "type": "general",
            "result": f"Direct response: {query}"
        }


    except Exception as e:

        logging.exception("Agent error")

        return {
            "type": "error",
            "result": str(e)
        }
````
